```python
class Baseball_Team:
    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw
    
    def calc_win_rate(self):
        return round(self.win / (self.win + self.lose), 3)
    
    def show_team_result(self):
        rate = self.calc_win_rate()
        print(f"{self.name:<10}{self.win:>4}{self.lose:>6}{self.draw:>6}{rate:>7.3f}")
```

```python
teams = [
    Baseball_Team("Archers", 77, 64, 2),
    Baseball_Team("Meteors", 71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks", 70, 70, 3),
    Baseball_Team("Pythons", 68, 73, 2),
    Baseball_Team("Owls", 59, 82, 2),
]

print(f"{'team':<10}{'win':>4}{'lose':>6}{'draw':>6}{'rate':>6}")

for team in teams:
    team.show_team_result()
```
