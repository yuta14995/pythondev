## 機械学習の流れ

### データを読み込む

```python
from sklearn import datasets
irisdata = datasets.load_iris()
```

```python
print(irisdata.DESCR)
```

```python
irisdata.data
```

```python
irisdata.target
```

```python
print(irisdata.data.shape)
print(irisdata.target.shape)
```

```python
print(irisdata.feature_names)
print(irisdata.target_names)
```

```python
X = irisdata.data
y = irisdata.target
```

### データの前処理を行なう

```python
# irisdata.dataを標準化
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X)
X = scaler.transform(X)
X
```

### データを訓練データとテストデータに分ける

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, test_size=0.2, shuffle=True)
```

### モデルを作成して訓練する



```python
from sklearn.svm import SVC
classifier = SVC(kernel="linear")
```

```python
classifier.fit(X_train, y_train)
```

### 期待する性能が出たかを評価する

#### テストデータで分類を実行する

```python
y_pred = classifier.predict(X_test)
y_pred
```

```python
y_test
```

#### モデルの性能を測る

##### `accuracy_score()`

```python
from sklearn import metrics
metrics.accuracy_score(y_test, y_pred)
```

##### 混同行列

```python
metrics.confusion_matrix(y_test, y_pred)
```

##### `classification_report()`

```python
print(metrics.classification_report(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import classification_report

# 例: y_test, y_pred がある前提で classification_report を辞書で取得
report = classification_report(y_test, y_pred, output_dict=True)
print(report)

# クラス名（ラベル）
labels = [key for key in report.keys() if key.isdigit()]  # 数値クラスだけ取得
metrics_names = ['precision', 'recall', 'f1-score']

# 各メトリクスごとの値を抽出
values = {metric: [report[label][metric] for label in labels] for metric in metrics_names}

# グラフの描画
x = np.arange(len(labels))  # x軸の位置
width = 0.25  # 棒の幅

fig, ax = plt.subplots(figsize=(8, 5))
for i, metric in enumerate(metrics_names):
    ax.bar(x + i * width, values[metric], width, label=metric)

# 軸ラベルとタイトル
ax.set_xlabel('Class Label')
ax.set_ylabel('Score')
ax.set_title('Classification Report Metrics')
ax.set_xticks(x + width)
ax.set_xticklabels(labels)
ax.set_ylim(0, 1.1)
ax.legend()

plt.tight_layout()
plt.show()
```

## 実習1：手書き数字の画像の分類

#### データを読み込む

```python
from sklearn import datasets
digits = datasets.load_digits()
```

```python
print(digits.DESCR)
```

```python
X = digits.data
y = digits.target
```

```python
print(X.shape)
print(y.shape)
```

```python
print(X[0])
print(y[0:50])
```

```python
import matplotlib.pyplot as plt
#画像の枠組みを作る
fig = plt.figure()

for i, x in enumerate(X[0:10], 0):#インデックスと最初の10個の画像データを取る。
    sp = fig.add_subplot(2, 5, (i + 1))#２×５に画像を並べるよ。
    sp.imshow(x.reshape(8, 8), cmap = "gray")#８×８に並べるよ。
```

### データを訓練データとテストデータに分ける

```python
X_train = X[:1200]
X_test  = X[1200:]
y_train = y[:1200]
y_test  = y[1200:]
```

### モデルを作成して訓練する

```python
from sklearn.svm import SVC
classifier = SVC(kernel="linear")
classifier.fit(X_train, y_train)
```

### 期待する性能が出たかを評価する

```python
y_pred = classifier.predict(X_test)
```

```python
from sklearn import metrics
print(metrics.confusion_matrix(y_test, y_pred))
```

```python
print(metrics.classification_report(y_test, y_pred))
```

## 実習2：住宅価格の分析と予測

### データを読み込む

```python
import pandas as pd
# 訓練データ
df = pd.read_csv("./sample_data/california_housing_train.csv")
df
```

```python
# テストデータ
df_test = pd.read_csv("./sample_data/california_housing_test.csv")
df_test
```

```python
# 訓練データ
X_train = df[df.columns[:-1]]#columnsで列名を取得し、最後の列名以外を取得
y_train = df['median_house_value']

# テストデータ
X_test = df_test[df_test.columns[:-1]]
y_test = df_test['median_house_value']
```

### モデルを作成して訓練する

```python
from sklearn import linear_model
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

### 期待する性能が出たかを評価する

```python
y_pred = model.predict(X_test)
```
