```python
result = []
books = dict.fromkeys(list("abcdefghijklmnopqrstuvwz"), 0)
print(books)

#単語を入力して格納する。
while True:
    word = input("英単語を入力してください：")
    if word == "":
        break
    result.append(word)

#各単語を順番に取り出してbooksから検索してカウントする。
for word in result:
    for w in word: 
        if w in books:
            books[w] += 1

#表示させる。
print(f"入力した英単語：{sorted(result)}")
for key, count in books.items():
    if count > 0:
        print(f"{key}が{count}個ありました")
```
